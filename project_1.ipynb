```python
# Simple Calculator Project
# This calculator performs different mathematical operations
# and also stores calculation history.

print("Welcome to Simple Calculator")

# Empty list to store calculation history
history = []

# Counter to count successful calculations
count = 0


# Keep the calculator running until the user chooses Exit
while True:

    # ---------------- MENU ----------------
    print("\nChoose an operation:")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    print("5. Modulus (%)")
    print("6. Power (**)")
    print("7. Exit")
    print("8. View History")
    print("9. Clear History")
    print("10. Total Calculations")
    print("11. Reset Calculator")

    # Take operation choice from the user
    choice_operation = input("Enter your operation (1-11): ")


    # ---------------- EXIT ----------------
    if choice_operation == "7":
        print("Thank you for using the calculator. Goodbye!")
        break


    # ---------------- VIEW HISTORY ----------------
    if choice_operation == "8":

        # Check whether history is empty
        if len(history) == 0:
            print("No calculator history yet.")

        else:
            print("\n=== Calculator History ===")

            # Print every calculation stored in history
            for item in history:
                print(item)

        # Go back to the menu
        continue


    # ---------------- CLEAR HISTORY ----------------
    if choice_operation == "9":

        # Ask for confirmation
        confirm = input(
            "Are you sure you want to clear history? (yes/no): "
        )

        if confirm.lower() == "yes":
            history.clear()
            print("Calculator history cleared successfully!")

        else:
            print("History was not cleared.")

        # Go back to the menu
        continue


    # ---------------- TOTAL CALCULATIONS ----------------
    if choice_operation == "10":

        print("Total calculations performed:", count)

        # Go back to the menu
        continue


    # ---------------- RESET CALCULATOR ----------------
    if choice_operation == "11":

        # Ask for confirmation
        confirm = input(
            "Are you sure you want to reset everything? (yes/no): "
        )

        if confirm.lower() == "yes":

            # Delete all history
            history.clear()

            # Reset calculation counter
            count = 0

            print("Calculator reset successfully!")

        else:
            print("Calculator was not reset.")

        # Go back to the menu
        continue


    # ---------------- CHECK VALID OPERATION ----------------

    # Only choices 1-6 are mathematical operations
    if choice_operation not in ["1", "2", "3", "4", "5", "6"]:

        print("Invalid choice! Please choose between 1 and 11.")

        # Go back to menu
        continue


    # ---------------- ENTER NUMBERS ----------------

    # try is used to prevent the program from crashing
    # if the user enters something that is not a number.
    try:

        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

    except ValueError:

        print("Please enter valid numbers!")

        # Go back to menu
        continue


    # ---------------- ADDITION ----------------
    if choice_operation == "1":

        # Calculate addition
        result = num1 + num2

        # Display result
        print(num1, "+", num2, "=", result)

        # Store calculation in history
        history.append(f"{num1} + {num2} = {result}")

        # Increase calculation count
        count += 1


    # ---------------- SUBTRACTION ----------------
    elif choice_operation == "2":

        # Calculate subtraction
        result = num1 - num2

        # Display result
        print(num1, "-", num2, "=", result)

        # Store calculation in history
        history.append(f"{num1} - {num2} = {result}")

        # Increase calculation count
        count += 1


    # ---------------- MULTIPLICATION ----------------
    elif choice_operation == "3":

        # Calculate multiplication
        result = num1 * num2

        # Display result
        print(num1, "*", num2, "=", result)

        # Store calculation in history
        history.append(f"{num1} * {num2} = {result}")

        # Increase calculation count
        count += 1


    # ---------------- DIVISION ----------------
    elif choice_operation == "4":

        # Division by zero is not allowed
        if num2 == 0:

            print("Error: Division by zero is not allowed.")

        else:

            # Calculate division
            result = num1 / num2

            # Display result
            print(num1, "/", num2, "=", result)

            # Store calculation in history
            history.append(f"{num1} / {num2} = {result}")

            # Increase calculation count
            count += 1


    # ---------------- MODULUS ----------------
    elif choice_operation == "5":

        # Modulus by zero is not allowed
        if num2 == 0:

            print("Error: Modulus by zero is not allowed.")

        else:

            # Calculate remainder
            result = num1 % num2

            # Display result
            print(num1, "%", num2, "=", result)

            # Store calculation in history
            history.append(f"{num1} % {num2} = {result}")

            # Increase calculation count
            count += 1


    # ---------------- POWER ----------------
    elif choice_operation == "6":

        # Calculate power
        # Example: 2 ** 3 = 8
        result = num1 ** num2

        # Display result
        print(num1, "**", num2, "=", result)

        # Store calculation in history
        history.append(f"{num1} ** {num2} = {result}")

        # Increase calculation count
        count += 1
```
